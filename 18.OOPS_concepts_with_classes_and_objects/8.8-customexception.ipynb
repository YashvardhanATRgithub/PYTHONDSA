### Custom exception (Raise and Throw an exception)

```python
# Step 1: Define a base custom error class
class Error(Exception):
    """Base class for all custom exceptions."""
    pass

# Step 2: Define a specific custom exception class for DOB validation
class dobException(Error):
    """Raised when the DOB does not fall in the valid range (20-30 years)."""
    pass
```

```python
# Step 3: Ask the user to enter their year of birth
year = int(input("Enter your year of birth (e.g., 1998): "))

# Step 4: Calculate the age based on the current year (2024 is hardcoded here)
age = 2024 - year

# Step 5: Use try-except to handle valid and invalid age scenarios
try:
    if 20 <= age <= 30:
        print("The age is valid. You can apply for the exam.")
    else:
        # Raise custom exception if age is not within the range
        raise dobException
except dobException:
    # Handle the exception gracefully
    print("Sorry, your age should be between 20 and 30 to be eligible.")
```
